```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
# Educación (Módulo 03)
df3 = pd.read_stata('enaho01a_2020_300.dta')
df3.head()
```

```python
# Samaria (Módulo 34) - Pobreza Monetaria (Enfoque de Ingresos y Gastos)
df34 = pd.read_spss('Sumaria-2020.sav')
df34.head()
```

```python
# Sustraccion de columnas de interes
educacion = df3[['mes','conglome', 'vivienda', 'hogar', 'p203', 'p301a', 'p208a', 'p207']]
monetario = df34[['MES','CONGLOME', 'VIVIENDA', 'HOGAR', 'POBREZA', 'MIEPERHO', 'DOMINIO', 'ESTRATO', 'FACTOR07']]

# convertimos a minuscula los titulos de la tabla monetario
monetario.columns = monetario.columns.str.lower()

# convierto a csv para trabajar en power bi
educacion.to_csv('educacion.csv',index=False, encoding='utf-8-sig')
monetario.to_csv('monetario.csv',index=False, encoding='utf-8-sig')
```

```python
print(educacion.columns)
print(monetario.columns)
```

```python
# renombro las columnas
educacion.rename(columns={'p203':'parentesco','p301a':'nivel_educativo','p208a':'edad_jefe','p207':'sexo_jefe'}, inplace=True)
monetario.rename(columns={ 'mieperho':'tamano_hogar'}, inplace=True)

educacion.columns,monetario.columns
```

```python
# discretizo sexo del jefe del hogar
educacion['sexo_jefe'] = educacion['sexo_jefe'].apply(lambda x: 'Hombre' if x==1 else 'Mujer')
educacion['sexo_jefe']
```

```python
# filtrar solo los jefes de Hogar
jefes = educacion[educacion['parentesco'] == 1].copy()
jefes.head()
```

```python
# Verifico categoria de pobreza en la tabla monetario
monetario['pobreza'].value_counts()
```

```python
# Crear variable binaria de pobreza
monetario['pobre_binario'] = monetario['pobreza'].apply(lambda x: 1 if x=='Pobre No Extremo' or x=='Pobre Extremo' else 0)
monetario.head()
monetario['pobre_binario'].value_counts()
```

```python
# juntamos ambas tablas por 'conglome', 'vivienda', 'hogar'
ENAHO = pd.merge(
    educacion, monetario,
    on = ['mes','conglome', 'vivienda', 'hogar'],
    how = 'inner'
)
```

```python
# convierto a csv para trabajar en power bi
educacion.to_csv('educacion.csv',index=False, encoding='utf-8-sig')
monetario.to_csv('monetario.csv',index=False, encoding='utf-8-sig')
ENAHO.to_csv('ENAHO.csv',index=False, encoding='utf-8-sig')
```
